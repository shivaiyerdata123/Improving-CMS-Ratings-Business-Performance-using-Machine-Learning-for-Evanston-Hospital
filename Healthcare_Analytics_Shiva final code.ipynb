# 💼 Business Understanding

The **Centers for Medicare & Medicaid Services (CMS)** is a U.S. federal agency responsible for administering the nation’s largest public healthcare programs — **Medicare** and **Medicaid**. In addition to program administration, CMS plays a critical role in collecting and analyzing healthcare data, publishing performance benchmarks, and implementing measures to improve quality while preventing fraud and abuse across the healthcare system.

CMS’s overarching mission is to:
- Improve **quality of care**
- Expand **access to healthcare coverage**
- Enhance **patient health outcomes**
- Reduce **overall healthcare costs**, including hospital expenditures and insurance premiums

---

To promote **transparency** and empower patients to make **informed healthcare decisions**, CMS evaluates and rates hospitals across the United States using a **five-star rating system**. These ratings allow patients, caregivers, and insurance providers to compare hospitals based on standardized quality and performance metrics.

Because CMS ratings directly influence **patient choice**, **payer preferences**, and ultimately **hospital revenue and profitability**, it is imperative for healthcare providers to understand the **key drivers behind these ratings** and focus their improvement efforts accordingly.

---

Evanston Hospital approached our team to understand why **profit growth has stalled** despite significant investments aimed at improving service quality.

### Key Insights from Management Discussions

- Over the last **six quarters**, Evanston Hospital’s **net profit has declined by approximately 7%**  
- The number of **patients and insurance companies** choosing Evanston Hospital has decreased by **~5% annually over the past five years**  
- In contrast, the overall demand for **private healthcare providers in the region has increased**, indicating patient migration to competing hospitals  
- A **patient and insurer survey conducted 10 months ago** identified Evanston’s **CMS rating of 3 stars** as a major deterrent  
- Patients and payers showed a clear preference for hospitals with **higher CMS ratings**  
- In response, hospital leadership invested heavily in:
  - New medical equipment  
  - Specialized staff training  
  - Advanced facilities such as a **high-risk obstetrics laboratory** and an **Infant Special Care Unit**
- Despite these initiatives, **the CMS rating remained unchanged**
- Further discussions revealed that:
  - Management lacked clarity on **which specific factors most strongly influence CMS ratings**
  - Investment decisions were largely driven by **assumptions rather than data-backed insights**

```python
# creating user-defined function for checking nulls
def null_inspection(df):
    
    Null_results_Count = pd.DataFrame(df.isnull().sum().sort_values(ascending=False), 
                              columns=['Null_Count'])
    
    Null_results_Percentage = pd.DataFrame(round((df.isnull().sum()/len(df))*100, 2).sort_values(ascending=False),
                                   columns=['Null_Percentage'])
    
    return pd.concat([Null_results_Count, Null_results_Percentage], axis = 1)
```

## 📐 Final Regression Equation (OLS Model)

Based on the Ordinary Least Squares (OLS) regression analysis, the best-fitted linear equation for predicting **Hospital Overall Rating** is:

\[
\begin{aligned}
\text{Hospital Overall Rating} =\;& 3.0866 \\
&- 0.0874 \times \text{Mortality\_MORT\_30\_HF} \\
&- 0.1212 \times \text{Mortality\_MORT\_30\_PN} \\
&- 0.3133 \times \text{Safetyofcare\_PSI\_90\_SAFETY} \\
&+ 0.0649 \times \text{PatientExperience\_H\_COMP\_1} \\
&+ 0.1147 \times \text{PatientExperience\_H\_COMP\_3} \\
&+ 0.0844 \times \text{PatientExperience\_H\_COMP\_7} \\
&+ 0.1091 \times \text{PatientExperience\_H\_HSP\_RATING} \\
&- 0.1345 \times \text{Mortality\_National\_Comparison\_Below} \\
&- 0.0696 \times \text{Mortality\_National\_Comparison\_Same} \\
&- 0.1246 \times \text{PatientExperience\_National\_Comparison\_Below} \\
&- 0.1098 \times \text{PatientExperience\_National\_Comparison\_Same} \\
&- 0.0770 \times \text{EffectivenessOfCare\_National\_Comparison\_Below} \\
&- 0.3752 \times \text{Readmission\_National\_Comparison\_Below} \\
&- 0.1930 \times \text{Readmission\_National\_Comparison\_Same} \\
&- 0.0753 \times \text{Hospital\_Type\_Critical\_Access}
\end{aligned}
\]

### 📌 Interpretation
Each coefficient represents the expected change in the hospital’s overall rating for a one-unit increase in the corresponding predictor, holding all other variables constant.  
Positive coefficients indicate factors that improve hospital ratings, while negative coefficients indicate factors that reduce ratings.  
Categorical variables encoded as dummy variables are interpreted relative to their respective reference categories.

#### Equation for best fitted line is:

Rating = 3.0866 + -0.0874 x Mortality_MORT_30_HF + -0.1212 x Mortality_MORT_30_PN + -0.3133 x Safetyofcare_PSI_90_SAFETY + 0.0649 x PatientExperience_H_COMP_1 + 0.1147 x PatientExperience_H_COMP_3 + 0.0844 x PatientExperience_H_COMP_7 + 0.1091 x PatientExperience_H_HSP_RATING + -0.1345 x Mortality national comparison_Below + -0.0696 x Mortality national comparison_Same + -0.1246 x Patient experience national comparison_Below + -0.1098 x Patient experience national comparison_Same + -0.077 x Effectiveness of care national comparison_Below + -0.3752 x Readmission national comparison_Below + -0.193 x Readmission national comparison_Same + -0.0753 x Hospital Type_Critical_Access

The analysis of CMS Hospital Compare data for **Evanston Hospital (Provider ID: 140010)** reveals that the hospital’s **declining profitability is not driven by weak clinical capability**, but by **operational inefficiencies and inconsistent patient experience**, which collectively suppress its CMS star rating.

- High **30-day readmission rates**, particularly for COPD, are a major source of CMS penalties.
